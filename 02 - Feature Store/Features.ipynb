```python
import pandas as pd
```

```python
def create_feature(feature_name, dt_inicio, dt_fim):

    dates = [i.strftime("%Y-%m-%d") for i in pd.date_range(dt_inicio, dt_fim, freq='MS')]

    with open(f"{feature_name}.sql") as file_open:
        query = file_open.read()

    for i in dates:
        print(i)
        df = spark.sql(query.format(date=i))

        (df.write
        .format("delta")
        .mode("overwrite")
        .option("replaceWhere", f"dtRef = '{i}'")
        .saveAsTable(f"workspace.tmw_trabalhofinal.{feature_name}"))    
```

```python
create_feature('fs_pontos', '2025-01-01', '2026-06-01')
```

```python
create_feature('fs_cursos', '2025-01-01', '2026-06-01')
```
